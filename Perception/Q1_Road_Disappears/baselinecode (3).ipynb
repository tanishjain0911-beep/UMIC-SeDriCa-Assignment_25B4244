```python
import cv2
import numpy as np

frame = cv2.imread("missing/018.png")

v = 170
brightness = np.sum(frame[v, :, :3], axis=1)

for name, start, end in [
    ("LEFT", 0, 240),
    ("RIGHT", 240, 480)
]:
    values = brightness[start:end]
    unique_values = np.unique(values)[::-1]

    print(f"\n{name} HALF")

    for rank, value in enumerate(unique_values[:10], start=1):
        u_values = np.where(values == value)[0] + start
        print(f"Rank {rank}: Brightness = {value}, u = {u_values}")
```

```python
#from the above analysis for multiple frames, we get to know that the real boundary has a brighntess of 686,
# and the beam(false boundary) has a brightness of 542.
# So whenever we calculate the brightest pixel , we will set the threshold to be 600.
```

```python
def center(frame):
    vn = 260
    vf = 170
    threshold = 600

    # ---------- FAR ROW ----------
    far = np.sum(frame[vf, :, :3], axis=1)

    left_far = far[:240]
    right_far = far[240:]

    lf = np.argmax(left_far)
    rf = np.argmax(right_far) + 240

    # ---------- NEAR ROW ----------
    near = np.sum(frame[vn, :, :3], axis=1)

    left_near = near[:240]
    right_near = near[240:]

    ln = np.argmax(left_near)
    rn = np.argmax(right_near) + 240

    # Check whether boundaries are bright enough
    if left_near[ln] < threshold or right_near[rn - 240] < threshold:
        un = None
    else:
        un = (ln + rn) / 2

    if left_far[lf] < threshold or right_far[rf - 240] < threshold:
        uf = None
    else:
        uf = (lf + rf) / 2

    return un, uf
```

```python
import os
import cv2

folder = "clear"

files = sorted(os.listdir(folder))

for file in files:
    if file.endswith(".png"):
        path = os.path.join(folder, file)
        frame = cv2.imread(path)

        centern, centerf = center(frame)

        print(file, centern, centerf)
```

```python
#Image region:
#The baseline examines only the two horizontal rows specified in the assignment:
#\(v_n=260\) (near row) and \(v_f=170\) (far row). At each row, the image is divided 
#into a left half (\(u<240\)) and a right half (\(u\geq240\)). 
#The brightest pixel in each half is selected as the corresponding left or right boundary candidate.
```

```python
#Brightness threshold:
#Pixel brightness is calculated as the sum of the three colour-channel values,
#\[B=R+G+B\]
#and a candidate boundary is considered valid only if its maximum brightness exceeds the 
#selected threshold of 600. 
#If the brightest candidate on either side is below this threshold, 
#that boundary is treated as unknown (None), rather than assigning a boundary without sufficient
#visual evidence.
```

```python
#Lane-centre calculation:
#When both boundary candidates are available, the lane centre is calculated as their midpoint:
#\[u_n=\frac{u_{L,n}+u_{R,n}}{2}\]\[u_f=\frac{u_{L,f}+u_{R,f}}{2}\]
#If either boundary is unavailable, the corresponding lane-centre estimate is marked as unknown.
```

```python
#Assumptions:
#The baseline assumes that the lane boundaries are approximately separated into the left 
#and right halves of the 480-pixel-wide image and that the brightest pixel
#in each half corresponds to a road-boundary marking.
```
